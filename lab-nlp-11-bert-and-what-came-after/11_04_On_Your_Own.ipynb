# 11_04 On your own: Kittiwake's answer desk

Kittiwake's support team wants a first-line answer desk: a customer's question in, the exact words from a
service notice out, or nothing at all when the notices do not say. A wrong answer about a price or an outage
is worse than no answer, because the customer acts on it.

**The brief**

- `desk_questions.json` holds 15 questions customers have asked. Some are answered by a notice; some are not.
- Write `answer(question)`: it returns a span copied from one of `bertlab.notices()`, or `""` to decline.
- Save every answer to `out/kittiwake_answers.json`, as `{question: answer}`.
- **Bar:** at least **12 of the 15** right, where declining a question the notices do not answer counts as right.

Everything you need is in the last notebook and in Lab 02: `bertlab.read()`, its `logit`, and a way to pick
the notice before the reader sees it. The page for this notebook has a planning prompt.
