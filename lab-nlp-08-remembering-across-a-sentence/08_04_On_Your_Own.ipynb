# 08_04 On your own: a headline classifier that reads in order

**The brief.** A news aggregator files incoming headlines under four categories: business, entertainment,
health, and science and technology. Its next project flags headlines whose meaning turns on word order, so the
team wants the classifier built on a network that reads in order, and retrained nightly on a small machine.
Build it:

- a model with a recurrent layer (`nn.RNN`, `nn.LSTM` or `nn.GRU`), trained on `X_train`, `y_train` from
  `headlines.load_split()`;
- **training must finish within five minutes** in this session; time it with `time.time()`;
- save it with `headlines.save_model(model, "out/headline_model.pt", **kwargs)`, where `kwargs` are the
  arguments that rebuild it (for example `cell="lstm", hidden=128`);
- save a short report, `out/headline_report.json`, with your model's class and arguments, epochs, optimiser,
  learning rate, training seconds and your own test scores.

**The bar.** On the 11,876 held-out headlines, which the checkpoint scores itself: macro-F1 at least
**0.895**. The four categories are balanced, so accuracy lands within a hair of macro-F1.

**Rules of the road.** `SequenceClassifier` takes `cell`, `bidirectional`, `pack`, `pool`, `embed_dim`,
`hidden`, `layers` and `dropout`; `headlines.train` takes `epochs`, `optimizer`, `batch_size` and `clip`. You
may also add your own class to `headlines.py` (a class defined in this notebook cannot be loaded by the
checkpoint, because it exists only in this kernel). Fix a seed.

There is no worked example in this notebook. `check_on_your_own()` loads your saved model and scores it
exactly as the checkpoint will.
