# 10_04 On your own: a transformer that sorts headlines

**The brief.** Kittiwake Mobile's press office watches the news and wants every incoming headline routed to one
of four desks: business, entertainment, health, and science and technology. Build the router as a transformer:

- train on `X_train`, `y_train` from `attnlab.load_split()`: 49,376 real headlines;
- the model must use **attention**: at least one `nn.MultiheadAttention` layer, for example through
  `attnlab.HeadlineTransformer` or your own class built from `attnlab.TransformerBlock`;
- the **training must finish within three minutes** in this session; time it with `time.time()`;
- save it with `attnlab.save_model(model, "out/headline_model.pt", **kwargs)`, where `kwargs` are the arguments
  that rebuild it (for example `layers=1`);
- save a short report, `out/headline_report.json`, with your model's class, settings, epochs, training seconds
  and your own test scores.

**The bar.** On the 10,000 held-out headlines, which the checkpoint scores itself: accuracy at least **0.88**
and macro-F1 at least **0.88**. The four desks are the same size, so the two numbers move together.

**Rules of the road.** Any class in `attnlab.py` that contains attention is fine, and so is a class you add to
that file (a class defined in this notebook cannot be loaded by the checkpoint, because it exists only in this
kernel). You may change the width `d`, heads, layers, feed-forward size, dropout, optimiser, learning rate,
epochs and batch size. Fix a seed.

There is no worked example in this notebook. `check_on_your_own()` loads your saved model and scores it
exactly as the checkpoint will.
