# 06_04 On your own: a neural network that routes Kittiwake's tickets

**The brief.** In Lab 03 you routed Kittiwake Mobile's support tickets to four departments with logistic
regression. Now build the router as a small **neural network** in PyTorch and see whether a hidden layer
earns its place.

- Train on the 600 labelled tickets in `data/kittiwake_tickets.csv` (`text`, `department`).
- Features: sentence embeddings from `neurontools.embed(texts)` (384 numbers per ticket, already computed
  in the background when your session started), or TF-IDF from scikit-learn turned into a dense tensor.
- The network: at least one hidden layer with an activation, then 4 outputs, trained with
  `nn.CrossEntropyLoss`.
- Hold back part of the labelled tickets as a **validation** set, measure macro-F1 on it, and use it to choose
  the hidden size and the number of training steps. Then retrain on everything.
- Predict a department for each of the 200 tickets in `data/kittiwake_tickets_unlabeled.csv`.

Write `out/ticket_predictions.csv` with columns `ticket_id,department`, and `out/network_report.json` with
at least `{"features": ..., "hidden": ..., "epochs": ..., "validation_macro_f1": ...}`.

The checkpoint scores your predictions against departments you have not seen, with a pass bar of 0.85
macro-F1. Two things to know: about 6 percent of Kittiwake's tickets were routed wrongly by the people who
labelled them, so even a perfect reader of the text cannot score much above 0.92; and Lab 03's logistic
regression reached about 0.92 on the same tickets, so a network is not guaranteed to beat it. Say in your
report whether yours did.

There is no worked example in this notebook. `check_on_your_own()` checks that your files are complete.
