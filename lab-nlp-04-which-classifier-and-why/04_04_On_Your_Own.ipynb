# 04_04 On your own: can twenty examples beat three thousand?

**The brief.** Kittiwake Mobile wants every app-store review labelled positive or negative. You have 3,000
labelled review sentences from Amazon, IMDB and Yelp, and only **20** labelled Kittiwake reviews
(`data/kittiwake_reviews_labelled.csv`). Label the other 380 (`data/kittiwake_reviews_unlabelled.csv`) and
write `out/review_sentiment.csv` with two columns, `review_id` and `label` (1 positive, 0 negative), one row per
review.

**How it is scored.** Macro-F1 on the reviews whose star rating is clear (1 or 2 stars negative, 4 or 5
positive; 3-star reviews are ignored), which you do not have. The bar is **0.92**.

**What you will run into.** A model trained on the 3,000 sentences alone scores about 0.77 on Kittiwake's
reviews. The words are different: roaming fees, eSIMs, 5G and towns that the Amazon, IMDB and Yelp
reviewers never wrote about. This is **domain shift**, and it is the most common reason a model that tested well
fails in production. Your 20 labelled reviews are the fix, if you can make the model listen to them.
