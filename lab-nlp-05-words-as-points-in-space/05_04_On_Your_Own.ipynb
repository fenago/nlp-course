# 05_04 On your own: what did this customer buy next?

**The brief.** The shop wants a "customers also bought" box. `part3_queries(test_baskets, known)` gives
you 500 questions: each is a query id and one product from a real invoice after 1 November 2011. For
each, recommend **ten different products**, not including the query product itself. You score a hit when
any of your ten was really in that invoice.

Write your answers to `out/recommendations.json` as `{"Q000": ["22386", ...ten codes...], ...}`.

**The bar.** A hit rate at 10 of at least **0.50**. Recommending the best-sellers to everyone scores about
0.34, so you need something that uses the query product.

**The rules.** Learn only from `train_baskets`, the invoices before November. The test invoices are in
the same data file, and nothing stops you reading them, but a recommender that has seen the answers is
not a recommender; the checkpoint cannot tell, so this one is on your honour.

**Tools.** Anything from 05_02 and 05_03: `cooccurrence(train_baskets)` for counts of what is bought
together, `product_vectors(train_baskets)` for the skip-gram vectors (already trained in the background),
or both, for example counts first and vectors to fill any gaps. Can you beat the counting baseline's
0.63?

There is no worked example. `check_on_your_own()` scores your file exactly as the checkpoint will.
