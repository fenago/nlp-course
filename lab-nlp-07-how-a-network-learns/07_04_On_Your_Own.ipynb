# 07_04 On your own: a newswire topic classifier

**The brief.** A news desk wants incoming Reuters newswires sorted into six topics (earn, acq, crude, trade,
money-fx, interest) as they arrive. Build the classifier:

- train any network you like on `X_train`, `y_train` from `newswire.load_split()`;
- the **training must finish within two minutes** in this session (the desk retrains nightly on a small
  machine); time it with `time.time()`;
- save it with `newswire.save_model(model, "out/newswire_model.pt", **kwargs)`, where `kwargs` are the
  arguments that rebuild it (for example `hidden=128`);
- save a short report, `out/newswire_report.json`, with your model's class, epochs, optimiser, learning rate,
  training seconds and your own test scores.

**The bar.** On the 2,144 held-out test newswires, which the checkpoint scores itself: accuracy at least
**0.85** and macro-F1 at least **0.70**. Macro-F1 is the average over the six topics, so a model that ignores
the small topics cannot pass by getting "earn" right.

**Rules of the road.** Any class in `newswire.py` is fine, and so is a class you add to that file (a class
defined in this notebook cannot be loaded by the checkpoint, because it exists only in this kernel). You may
change the embedding size, hidden size, pooling, optimiser, learning rate, epochs and batch size. Fix a seed.

There is no worked example in this notebook. `check_on_your_own()` loads your saved model and scores it
exactly as the checkpoint will.

```python
torch.manual_seed(0)
t = time.time()
model = None     # YOUR CODE HERE: build a model, and train it within the time budget
train_seconds = time.time() - t
print(f"trained in {train_seconds:.0f} s")
```
