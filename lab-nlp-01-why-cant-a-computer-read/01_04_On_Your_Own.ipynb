# 01_04 On your own: the preprocessing step for Kittiwake's inbox

**The brief.** Kittiwake Mobile's support team wants every incoming ticket turned into something the rest
of their tools can use. You are building the first step. For each of the 600 tickets in
`data/kittiwake_tickets.csv`, produce one JSON object with:

- `ticket_id`: as in the file;
- `tokens`: the ticket's tokens, as strings;
- `lemmas`: the words that matter, as lowercase lemmas, with punctuation and stop words removed, and with
  **every negation kept as `not`**;
- `entities`: the named entities as `[text, label]` pairs, found on the **raw** ticket text.

Write them to `out/tickets_preprocessed.jsonl`, one object per line.

**Things the tickets will do to you.** About a fifth arrive by SMS, lowercased and without apostrophes:
*cant*, *dont*, *wont*. Some mention the same department's words in other departments' tickets. None of
that changes the brief.

**Tools.** Anything from the last three notebooks. spaCy does most of this in one pass (its lemmatizer
already turns `n't` into `not`, and it knows what *cant* means); NLTK does it in several. Either is fine.
What you may not do is lose a "not".

There is no worked example in this notebook. `check_on_your_own(sample=50)` checks the first 50 tickets
quickly while you work; `check_on_your_own()` checks all 600, as the checkpoint will.
